````python
import inspect
import os
import subprocess
import sys
from pathlib import Path

repo = Path.cwd() / 'drosophila-pd-flygym'
if not (repo / 'pyproject.toml').is_file():
    subprocess.run(['git', 'clone', 'https://github.com/TanVi3001/drosophila-pd-flygym.git', repo.name], check=False)
if (repo / 'pyproject.toml').is_file():
    os.chdir(repo)
    try:
        import flygym
        from flygym import Simulation
        from flygym.compose import BlocksTerrainWorld, FlatGroundWorld, MixedTerrainWorld, NeuroMechFly
        from flygym.rendering import Renderer
        objects = [NeuroMechFly, FlatGroundWorld, BlocksTerrainWorld, MixedTerrainWorld, Simulation, Renderer]
        lines = ['# FlyGym 2.1.0 API Summary', '']
        for item in objects:
            lines.extend([f'## {item.__name__}', '', f'`{item.__module__}.{item.__name__}`', '', '```python', str(inspect.signature(item)), '```', ''])
        summary = '\n'.join(lines)
        Path('docs/v2/flygym_adapter/colab_api_summary.md').write_text(summary, encoding='utf-8')
        print(summary)
    except Exception as exc:
        print('API inspection failed:', type(exc).__name__, exc)
else:
    print('Repository checkout is unavailable.')
````
